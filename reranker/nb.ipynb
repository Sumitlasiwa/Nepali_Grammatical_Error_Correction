```python
import json
from typing import List

input_path = "./valid.jsonl"
output_path = "./valid_clean.jsonl"

def normalize(s: str) -> str:
    s = s.replace("|", "।")
    s = s.replace(" ।", "।")
    s = s.replace(" ?", "?")
    return s.strip()


def load_jsonl(path: str) -> List[dict]:
    """Read jsonl and normalize candidate sentences"""

    records = []

    with open(path, "r", encoding="utf-8") as f:
        for line in f:

            dic = json.loads(line)

            source = normalize(dic["source"])
            reference = normalize(dic["reference"])

            candidates = []
            seen = set()

            for candidate in dic["candidates"]:

                text = normalize(candidate["text"])
                logprob = candidate["logprob"]

                # remove duplicates
                if text not in seen:
                    candidates.append({
                        "text": text,
                        "logprob": logprob
                    })
                    seen.add(text)

            records.append({
                "source": source,
                "reference": reference,
                "candidates": candidates
            })

    return records


def save_jsonl(records: List[dict], path: str):
    """Save records to jsonl file"""

    with open(path, "w", encoding="utf-8") as f:
        for rec in records:
            f.write(json.dumps(rec, ensure_ascii=False) + "\n")

    print(f"Saved {len(records)} records → {path}")


# run pipeline
data = load_jsonl(input_path)
save_jsonl(data, output_path)
```

```python
import json
import random

input_file = "pairwise_valid.jsonl"          # your dataset
output_file = "valid.jsonl"

# read jsonl
with open(input_file, "r", encoding="utf-8") as f:
    data = [json.loads(line) for line in f]

# shuffle
random.shuffle(data)

# save shuffled jsonl
with open(output_file, "w", encoding="utf-8") as f:
    for row in data:
        f.write(json.dumps(row, ensure_ascii=False) + "\n")

print(f"Shuffled {len(data)} records → {output_file}")
```
